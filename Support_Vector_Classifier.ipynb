```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.datasets import make_regression
x,y = make_regression(n_samples =1000,n_features = 2, n_targets =1, noise = 3)
```

```python
x
```

```python
y
```

```python
sns.scatterplot(x = pd.DataFrame(x)[0], y = pd.DataFrame(x)[1])
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size =0.3,random_state=1)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.svm import SVR
svr = SVR(kernel='linear')
svr.fit(x_train,y_train)
```

```python
svr.fit(x_train,y_train)
```

```python
svr.coef_
```

```python
y_pred = svr.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
# hyperparameter tunong with SVR
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR

param_svr = {'C': [0.1, 1, 10, 100],
             'gamma': [0.1, 0.01, 0.001],
             'kernel': ['linear', 'rbf']}

gridsearch_svr = GridSearchCV(SVR(), param_grid=param_svr, scoring='r2', cv=5, verbose=3)
```

```python
gridsearch_svr.fit(x_train, y_train)

print("Best parameters for SVR:", gridsearch_svr.best_params_)
print("Best cross-validation R2 score for SVR:", gridsearch_svr.best_score_)

y_pred_svr_tuned = gridsearch_svr.best_estimator_.predict(x_test)
r2_score_svr_tuned = r2_score(y_test, y_pred_svr_tuned)
print("R2 score on test set with tuned SVR:", r2_score_svr_tuned)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Define parameters for GridSearchCV (assuming these are the desired parameters)
param = {'C': [0.1,0.2,1,2,3,10,50,100],
         'gamma': [1,0.1,0.2,0.001,0.003],
         'kernel':['linear']}

# Initialize and fit GridSearchCV for SVC
gridsearch = GridSearchCV(SVC(), param_grid = param, scoring='accuracy', cv=5, verbose = 3)
gridsearch.fit(x_train,y_train)

print("Best parameters for SVC:", gridsearch.best_params_)
print("Best cross-validation accuracy for SVC:", gridsearch.best_score_)
```

```python
from sklearn.metrics import accuracy_score
from sklearn import svm
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.datasets import make_classification
import numpy as np

# Re-create x, y for classification and df
x_cls,y_cls = make_classification(n_samples=1000, n_features=2, n_classes=2, n_clusters_per_class=2, n_redundant=0)
df = pd.DataFrame(np.c_[x_cls, y_cls], columns=['feature_0', 'feature_1', 'target'])

# Ensure df1 is defined and has polynomial features from df
df1 = df.copy()
df1["x1Square"] = df1["feature_0"]**2
df1["x2Square"] = df1["feature_1"]**2
df1["x1x2"] = df1["feature_0"]*df1["feature_1"]

# Extract x and y from the prepared df1 for the SVC model
x = df1[["x1Square", "x2Square", "x1x2"]]
y = df1["target"]

# Re-define x_train, x_test, y_train, y_test
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size =0.3,random_state=1)

# Re-define and fit clf to ensure it's available in this cell's context
clf = svm.SVC(kernel='linear')
clf.fit(x_train, y_train)

y_pred = clf.predict(x_test)
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

# Define parameters for GridSearchCV with a polynomial kernel
param_poly = {'C': [0.1, 1, 10, 100],
              'gamma': [1, 0.1, 0.01, 0.001],
              'degree': [2, 3, 4], # Degree is specific to polynomial kernel
              'kernel': ['poly']}

# Initialize and fit GridSearchCV for Polynomial SVC
gridsearch_poly = GridSearchCV(SVC(), param_grid = param_poly, scoring='accuracy', cv=5, verbose = 3)
gridsearch_poly.fit(x_train,y_train)

print("Best parameters for Polynomial SVC:", gridsearch_poly.best_params_)
print("Best cross-validation accuracy for Polynomial SVC:", gridsearch_poly.best_score_)
```
